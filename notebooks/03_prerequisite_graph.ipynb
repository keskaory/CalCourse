# Prerequisite Graph
This notebook models Berkeley course prerequisites as a directed graph. Each edge points from a prerequisite course to the course it unlocks.

## 1. Load Prerequisite Data

```python
import pandas as pd
import networkx as nx

prereqs = pd.read_csv(
    "../data/processed/prerequisites_fall_2026.csv"
)

prereqs.head()
```

## 2. Create Course Labels

```python
prereqs["course"] = (
    prereqs["subject"] + " " + prereqs["course_number"].astype(str)
)

prereqs["prerequisite"] = (
    prereqs["prereq_subject"] + " " + prereqs["prereq_number"].astype(str)
)

prereqs.head()
```

## 3. Build Directed Graph

```python
G = nx.DiGraph()

for _, row in prereqs.iterrows():
    G.add_edge(
        row["prerequisite"],
        row["course"]
    )
```

```python
G.number_of_nodes(), G.number_of_edges()
```

```python
list(G.predecessors("AEROENG 10"))
```

```python
list(G.successors("COMPSCI 61B"))[:20]
```

### Graph Observations
- Each node represents a Berkeley course.
- Directed edges represent prerequisite relationships. 
- Predecessors of a course represent its parsed prerequisites.
- Successors of a course represent courses that the prerequisite helps unlock.
